```python
import numpy as np
import pickle
import matplotlib.pyplot as plt
import os

import matplotlib
plt.rcParams.update({ "text.usetex": True, "font.family": "serif" })
font = {'size': 16}
matplotlib.rc('font', **font)
fmt = 'pdf'
%matplotlib inline
```

```python
def avg_var_comp(data, metric):
    nepochs = len(data)
    vsize = np.array(data[0]['val'][metric]).size
    tseries = np.zeros((nepochs, vsize))
    for jj in range(nepochs):
        tseries[jj, :] = np.array(data[jj]['val'][metric])
        
    avg = np.mean(tseries, axis=0)
    std = np.std(tseries, axis=0)
    return avg, std
```

```python
def fig_plotter(avg, std, cval, label, metric):    
    if metric=='recall':
        epochs = 10 * np.arange(1,avg.size+1)
        plt.errorbar(epochs, avg, std, c=cval, ls='--', label=label)      
    else:
        epochs = 10 * np.arange(1,avg[::10].size+1)
        plt.errorbar(epochs, avg[::10], std[::10], c=cval, ls='--', label=label)  
    
```

```python

MAIN_DIR = os.getcwd()
print(MAIN_DIR)

os.chdir('../model_stats/model_stats/')
cheb_w_10_edges = pickle.load(open("LGCN_CHEB_3_e64_nodes19360__10_10_4.pkl", "rb"))
cheb_w_1_edges = pickle.load(open("LGCN_CHEB_3_e64_nodes19360__1_10_4.pkl", "rb"))
sage_w_10_edges = pickle.load(open("LGCN_SAGE_3_e64_nodes19360__10_10_4.pkl", "rb"))
sage_w_1_edges = pickle.load(open("LGCN_SAGE_3_e64_nodes19360__1_10_4.pkl", "rb"))
lgc_w_10_edges = pickle.load(open("LGCN_LGC_3_e64_nodes19360__10_10_4.pkl", "rb"))
lgc_w_1_edges = pickle.load(open("LGCN_LGC_3_e64_nodes19360__1_10_4.pkl", "rb"))

os.chdir(MAIN_DIR)
os.chdir('../model_stats_amazon')
am_cheb_w_10_edges = pickle.load(open("LGCN_CHEB_3_e64_nodes18869__10_10_3.pkl", "rb"))
am_cheb_w_1_edges = pickle.load(open("LGCN_CHEB_3_e64_nodes18869__1_10_3.pkl", "rb"))
am_sage_w_10_edges = pickle.load(open("LGCN_SAGE_3_e64_nodes18869__10_10_3.pkl", "rb"))
am_sage_w_1_edges = pickle.load(open("LGCN_SAGE_3_e64_nodes18869__1_10_3.pkl", "rb"))
am_lgc_w_10_edges = pickle.load(open("LGCN_LGC_3_e64_nodes18869__10_10_3.pkl", "rb"))
am_lgc_w_1_edges = pickle.load(open("LGCN_LGC_3_e64_nodes18869__1_10_3.pkl", "rb"))

os.chdir(MAIN_DIR)
```

```python
spot_cheb_10_avg, spot_cheb_10_std = avg_var_comp(cheb_w_10_edges, metric='recall')
spot_sage_10_avg, spot_sage_10_std = avg_var_comp(sage_w_10_edges, metric='recall')
spot_lgc_10_avg, spot_lgc_10_std = avg_var_comp(lgc_w_10_edges, metric='recall')
fig_plotter(spot_cheb_10_avg, spot_cheb_10_std, 'k', label=r"Cheb", metric='recall')
fig_plotter(spot_sage_10_avg, spot_sage_10_std, 'b', label=r"Sage", metric='recall')
fig_plotter(spot_lgc_10_avg, spot_lgc_10_std, 'r', label=r"LGC", metric='recall')
plt.title('Recall at 300, 10 Negative Edges')
plt.ylabel('Recall at 300')
plt.xlabel('Epochs')
plt.legend()
plt.savefig("spotify_ten_edges.pdf", dpi=150, bbox_inches='tight')
```

```python
auc_spot_cheb_10_avg, auc_spot_cheb_10_std = avg_var_comp(cheb_w_10_edges, metric='roc')
auc_spot_sage_10_avg, auc_spot_sage_10_std = avg_var_comp(sage_w_10_edges, metric='roc')
auc_spot_lgc_10_avg, auc_spot_lgc_10_std = avg_var_comp(lgc_w_10_edges, metric='roc')
fig_plotter(auc_spot_cheb_10_avg, auc_spot_cheb_10_std, 'k', label=r"Cheb", metric='roc')
fig_plotter(auc_spot_sage_10_avg, auc_spot_sage_10_std, 'b', label=r"Sage", metric='roc')
fig_plotter(auc_spot_lgc_10_avg, auc_spot_lgc_10_std, 'r', label=r"LGC", metric='roc')
plt.title('AUC, 10 Negative Edges')
plt.ylabel('AUC')
plt.xlabel('Epochs')
plt.legend()
plt.savefig("spotify_ten_edges_auc.pdf", dpi=150, bbox_inches='tight')
```

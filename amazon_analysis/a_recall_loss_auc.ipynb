```python
import pickle
import os
from a_recall_loss_auc import *
import matplotlib
plt.rcParams.update({ "text.usetex": True, "font.family": "serif" })
font = {'size': 14}
matplotlib.rc('font', **font)
fmt = 'pdf'
```

```python
MAIN_DIR='/home/jovyan/code4/amazon_analysis/a_model_stats'
os.chdir(MAIN_DIR)

one_neg_dirc= MAIN_DIR+'/1_neg'
ten_neg_dirc= MAIN_DIR+'/10_neg'
```

```python
os.chdir(one_neg_dirc)
```

```python
cheb_w_1 = load_model_runs("CHEB",1)
sage_w_1 = load_model_runs("SAGE",1)
lgcn_w_1 = load_model_runs("LGC",1)

mean_recall_cheb_1, std_recall_cheb_1 = compute_recall(cheb_w_1)
mean_recall_sage_1, std_recall_sage_1 = compute_recall(sage_w_1)
mean_recall_lgcn_1, std_recall_lgcn_1 = compute_recall(lgcn_w_1)

mean_loss_cheb_1, std_loss_cheb_1 = compute_loss(cheb_w_1)
mean_loss_sage_1, std_loss_sage_1 = compute_loss(sage_w_1)
mean_loss_lgcn_1, std_loss_lgcn_1 = compute_loss(lgcn_w_1)

mean_auc_cheb_1, std_auc_cheb_1 = compute_AUC(cheb_w_1)
mean_auc_sage_1, std_auc_sage_1 = compute_AUC(sage_w_1)
mean_auc_lgcn_1, std_auc_lgcn_1 = compute_AUC(lgcn_w_1)
```

```python
plt.figure()
plot(mean_recall_sage_1, std_recall_sage_1, label="SAGE", color='b')
plot(mean_recall_cheb_1, std_recall_cheb_1, label="CHEB", color='r')
plot(mean_recall_lgcn_1, std_recall_lgcn_1, label="LGCN", color='g')
plt.legend()
plt.title("Amazon Data: Recall@392, 1 Negative Edge")
plt.xlabel("Epochs")
plt.ylabel("Recall")
plt.savefig("a_1_neg_recall.pdf", dpi=150, bbox_inches='tight')
```

```python
plt.figure()
plot(mean_loss_sage_1, std_loss_sage_1, label="SAGE", color='b')
plot(mean_loss_cheb_1, std_loss_cheb_1, label="CHEB", color='r')
plot(mean_loss_lgcn_1, std_loss_lgcn_1, label="LGCN", color='g')
plt.ylim(0.1,0.7)
plt.legend()
plt.title("Amazon Data: Loss, 1 Negative Edge")
plt.xlabel("Epochs")
plt.ylabel("Loss")
plt.savefig("a_1_neg_loss.pdf", dpi=150, bbox_inches='tight')
```

```python
plt.figure()
plot(mean_auc_sage_1, std_auc_sage_1, label="SAGE", color='b')
plot(mean_auc_cheb_1, std_auc_cheb_1, label="CHEB", color='r')
plot(mean_auc_lgcn_1, std_auc_lgcn_1, label="LGCN", color='g')
plt.legend()
plt.title("Amazon Data: AUC, 1 Negative Edge")
plt.xlabel("Epochs")
plt.ylabel("AUC")
plt.savefig("a_1_neg_auc.pdf", dpi=150, bbox_inches='tight')
```
